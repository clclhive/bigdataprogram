```python
def main_screen():
    print("\n도형 만들기")
    print("1. 도형 그리기")
    print("2. 종료")

def select_level():
    while True:
        try:
            level = int(input("프랙탈 반복횟수 입력 : "))
            if 1 <= level <= 4:
                return level
            else:
                print("1부터 4 사이의 숫자를 입력하세요.")
        except ValueError:
            print("숫자를 입력하세요.")

# 프랙탈 십자 도형 패턴 생성 함수
def pattern(n):
    base_pattern = [" * ", "***", " * "]  # n=1일 때 기본 패턴

    if n == 1:
        return base_pattern

    prev = pattern(n - 1)
    size = len(prev)
    result = []

    for row in base_pattern:
        for i in range(size):
            line = ""
            for star in row:
                if star == '*':
                    line += prev[i]
                else:
                    line += ' ' * size
            result.append(line)

    return result

# 도형 출력 함수
def draw(level):
    print(f"\n{level}단계 도형 출력\n")
    pattern_fin = pattern(level)
    for line in pattern_fin:
        print(line)

# 메인 메뉴 처리
def main_select_onetwo():
    while True:
        main_screen()
        num = input("번호를 입력하세요: ")
        if num == '1':
            level = select_level()
            draw(level)
        elif num == '2':
            print("프로그램을 종료합니다.")
            break
        else:
            print("올바른 번호를 입력하세요.")

main_select_onetwo()
```
